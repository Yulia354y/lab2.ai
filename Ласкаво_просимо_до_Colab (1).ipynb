```

# Бібліотека для роботи з числовими масивами
import numpy as np

# Бібліотека для побудови графіків
import matplotlib.pyplot as plt

# Бібліотека для створення нейронної мережі
from tensorflow import keras

# Клас для створення шару нейронів
from tensorflow.keras.layers import Dense
```

```

# Вхідні значення x
x = np.array([-10, -5, 0, 5, 10, 15, 20, 25], dtype=float)

# Вихідні значення y за формулою y = 23x + 32
y = np.array([-198, -83, 32, 147, 262, 377, 492, 607], dtype=float)

# Виводимо навчальні дані
print("Значення x:", x)
print("Значення y:", y)
```

```

# Створюємо послідовну модель
model = keras.Sequential()

# Додаємо один нейрон з одним входом
model.add(Dense(units=1, input_shape=(1,), activation='linear'))

# Виводимо структуру нейромережі
model.summary()
```

```

# Налаштовуємо процес навчання
model.compile(
    # Функція для обчислення помилки
    loss='mean_squared_error',

    # Оптимізатор для зменшення помилки
    optimizer=keras.optimizers.Adam(0.01)
)
```

```

# Навчаємо нейромережу протягом 3000 епох
history = model.fit(x, y, epochs=3000, verbose=0)

# Повідомлення про завершення навчання
print("Навчання завершено!")
```

```

# Будуємо графік зміни помилки
plt.plot(history.history['loss'])

# Назва графіка
plt.title("Графік навчання нейронної мережі")

# Підписи осей
plt.xlabel("Кількість епох")
plt.ylabel("Помилка (Loss)")

# Додаємо сітку
plt.grid(True)

# Виводимо графік
plt.show()
```

```

# Задаємо нове значення x для перевірки
test_x = np.array([[7.0]])

# Отримуємо прогноз нейромережі
prediction = model.predict(test_x, verbose=0)

# Виводимо прогноз
print("Прогноз нейромережі:", prediction[0][0])

# Обчислюємо правильний результат за формулою
print("Правильна відповідь:", 23 * 7 + 32)
```

```

# Отримуємо навчені параметри нейромережі
weights = model.get_weights()

# Виводимо вагу нейрона
print("Вага нейрона:", weights[0])

# Виводимо зміщення нейрона
print("Зміщення:", weights[1])
```
